```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

home_url = "https://thelibrary1994.com/collections/carol-christian-poell-the-repository"



def get_webdirver():
    options = webdriver.ChromeOptions()
    options.add_argument("--headless")
    options.add_argument("--disable-gpu")
    options.add_argument("--no-sandbox")
    options.add_argument("enable-automation")
    options.add_argument("--disable-infobars")
    options.add_argument("--disable-dev-shm-usage")
    driver = webdriver.Chrome(options=options)
    driver.implicitly_wait(2)
    return driver


driver = get_webdirver()
page_count = 8

result = set()
for page in range(1, page_count + 1):
    page_url = f"{home_url}?page={page}"
    driver.get(page_url)
    wait = WebDriverWait(driver, timeout=15)
    wait.until(EC.presence_of_element_located((By.CLASS_NAME, "custom-product-container")))
    product_links = driver.find_elements(By.CLASS_NAME, "full-unstyled-link")
    for product_link in product_links:
        product_url = product_link.get_attribute("href")
        result.add(product_url)
driver.quit()

result
```

```python
import json
import time
import uuid

def process_product(product_link):
    driver = get_webdirver()
    driver.get(product_link)
    wait = WebDriverWait(driver, timeout=15)
    wait.until(EC.presence_of_element_located((By.CLASS_NAME, "product__sku")))
    time.sleep(2)
    designer = driver.find_element(By.CLASS_NAME, "product-custom-vendor-link")
    product_name = driver.find_element(By.CLASS_NAME, "product__title")
    product_details = driver.find_element(By.CLASS_NAME, "product__description")
    segments = product_details.text.split("\n")

    model_number = segments[0].replace("MODEL: ", "") if len(segments) > 0 else ""
    colour = segments[1].replace("COLOUR: ", "") if len(segments) > 1 else ""
    material = segments[2].replace("MATERIAL: ", "") if len(segments) > 2 else ""
    description = segments[3] if len(segments) > 3 else ""
    sku = driver.find_element(By.CLASS_NAME, "product__sku").text.replace("SKU:", "").replace("\n", "")
    sku = f"{sku}_{uuid.uuid4()}"

    images = driver.find_elements(By.CLASS_NAME, "image-magnify-lightbox")
    image_urls = [image.get_attribute("src").split("?")[0] for image in images]

    with open(f"products/{sku.replace('/', '-')}.json", "a") as f:
        f.write(json.dumps({
            "url": product_link,
            "designer": designer.text,
            "product_name": product_name.text,
            "model_number": model_number,
            "colour": colour,
            "material": material,
            "description": description,
            "sku": sku,
            "images": image_urls
        }))

    driver.quit()




# product_link = 'https://thelibrary1994.com/products/xposed-vest-bag'
# process_product(product_link)


```

```python
product_links_to_process = result.copy()
failures = set()
successes = set()
```

```python
for product_link in product_links_to_process:
    try:
        print('processing', product_link)
        process_product(product_link)
        print('done')
        successes.add(product_link)
    except Exception as e:
        print('error', e)
        failures.add(product_link)


```
